```python
def kge(sim, obs):
    sim = np.array(sim)
    obs = np.array(obs)

    # Remove NaNs if necessary
    mask = ~np.isnan(sim) & ~np.isnan(obs)
    sim = sim[mask]
    obs = obs[mask]

    r = np.corrcoef(sim, obs)[0, 1]  # Correlation coefficient
    alpha = np.std(sim) / np.std(obs)  # Variability ratio
    beta = np.mean(sim) / np.mean(obs)  # Bias ratio

    kge_value = 1 - np.sqrt((r - 1)**2 + (alpha - 1)**2 + (beta - 1)**2)

    return kge_value, r, alpha, beta
```

```python
# The 'kge' function should be defined in a previous cell.
# predicted_monthly and true_monthly are assumed available.

# ------------------------------------------------
# Remove months with missing observations
# ------------------------------------------------
pred_filtered = predicted_monthly.copy()
true_filtered = true_monthly.copy()

results = []

# ------------------------------------------------
# Category-wise KGE
# ------------------------------------------------
for i in range(4):
    kge_val, r_val, alpha_val, beta_val = kge(
        pred_filtered[str(i)],
        true_filtered[str(i)]
    )

    results.append({
        'Category': f'Category {i}',
        'KGE': kge_val,
        'r': r_val,
        'alpha': alpha_val,
        'beta': beta_val
    })

# ------------------------------------------------
# Overall monthly KGE
# ------------------------------------------------
kge_val_monthly, r_val_monthly, alpha_val_monthly, beta_val_monthly = kge(
    pred_filtered["value"],
    true_filtered["value"]
)

results.append({
    'Category': 'Overall Monthly (All Months)',
    'KGE': kge_val_monthly,
    'r': r_val_monthly,
    'alpha': alpha_val_monthly,
    'beta': beta_val_monthly
})

# ------------------------------------------------
# Wet and Dry season definition
# Ecuadorian Andean páramo
# ------------------------------------------------
wet_months = [10, 11, 12, 1, 2, 3, 4, 5]
dry_months = [6, 7, 8, 9]

# Wet season
pred_wet = pred_filtered[
    pred_filtered["month"].isin(wet_months)
]["value"]

true_wet = true_filtered[
    true_filtered["month"].isin(wet_months)
]["value"]

kge_wet, r_wet, alpha_wet, beta_wet = kge(
    pred_wet,
    true_wet
)

results.append({
    'Category': 'Overall Monthly Wet Season',
    'KGE': kge_wet,
    'r': r_wet,
    'alpha': alpha_wet,
    'beta': beta_wet
})

# Dry season
pred_dry = pred_filtered[
    pred_filtered["month"].isin(dry_months)
]["value"]

true_dry = true_filtered[
    true_filtered["month"].isin(dry_months)
]["value"]

kge_dry, r_dry, alpha_dry, beta_dry = kge(
    pred_dry,
    true_dry
)

results.append({
    'Category': 'Overall Monthly Dry Season',
    'KGE': kge_dry,
    'r': r_dry,
    'alpha': alpha_dry,
    'beta': beta_dry
})

# Display results
df_kge = pd.DataFrame(results)
display(df_kge.round(2))

# Save to CSV
df_kge.round(2).to_csv("kge_Toreadora.csv", index=False)
```
